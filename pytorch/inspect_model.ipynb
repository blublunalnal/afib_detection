## Purpose

Check the structure between pytorch model and the original keras model.

## Deepbeat (tensorflow model)

```python
from tensorflow.keras.models import load_model

h5_file_path = r'C:\develop\afib_detection\keras\deepbeat.h5'

# 1. Load the model
model = load_model(h5_file_path, compile = False)
# print out layers of the model
model.summary()

```

## print out the activation function for each layer

```python
for i, layer in enumerate(model.layers):
    layer_name = layer.name
    layer_type = layer.__class__.__name__
    if hasattr(layer, 'activation'):
        activation = layer.activation.__name__
        print(f"Layer {i}: {layer_name} ({layer_type}) -> Activation: {activation}")
    else:
        print(f"Layer {i}: {layer_name} ({layer_type}) -> No activation attribute")

print("="*60)
```
